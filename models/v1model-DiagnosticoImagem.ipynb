```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np

img_size = 64 
input_shape = img_size * img_size 
imagens_path = '../data/raw/'

trais_ds = tf.keras.preprocessing.image_dataset_from_directory(
  imagens_path,
  validation_split=0.2,
  subset="training",
  seed=568,
  image_size=(img_size, img_size),
  batch_size=64,
  color_mode='grayscale'
  )
val_ds = tf.keras.preprocessing.image_dataset_from_directory(
  imagens_path,
  validation_split=0.2,
  subset="validation",
  seed=568,
  image_size=(img_size, img_size),
  batch_size=64,
  color_mode='grayscale'
)
normalization_layer = layers.Rescaling(1./255)
trais_ds = trais_ds.map(lambda x, y: (normalization_layer(x), y))
val_ds = val_ds.map(lambda x, y: (normalization_layer(x), y))

#para melhorar a performance, vamos usar cache e prefetch
trais_ds = trais_ds.cache().prefetch(buffer_size=tf.data.AUTOTUNE)
val_ds = val_ds.cache().prefetch(buffer_size=tf.data.AUTOTUNE)

model = models.Sequential([
    layers.Flatten(input_shape=(img_size, img_size)),
    
    layers.Dense(128, activation='relu'),
    
    layers.Dense(64, activation='relu'),
    
    layers.Dense(1, activation='sigmoid')
])

# 3. Compilando o Modelo
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Exibe a estrutura da rede
model.summary()

train = model.fit(trais_ds, validation_data=val_ds, epochs=20)
```
